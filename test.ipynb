```python
def isValid(s):
    stack = []
    mapping = { ")":"(",
                "}":"{",
                "]":"["}
    
    for char in s:
        if char in mapping.values():
            stack.append(char)
        elif char in mapping.keys():

            if not stack or mapping[char] != stack.pop():
                return False
    return not stack

isValid('(])')
```

```python
#counts successful parentheses in single string
def isValid(s):
    stack = []
    mapping = { ")":"(",
                "}":"{",
                "]":"["}
    counter = 0
    for char in s:
        if char in mapping.values():
            stack.append(char)
        elif char in mapping.keys():
            if not stack or mapping[char] == stack.pop():
                counter +=1

    return counter

isValid('([]){}')
```

```python
def isValid(n):
    
    mapping = { ")":"(",
                "}":"{",
                "]":"["}
    counter = len(n)
    for s in n:
        stack = []
        for char in s:
            if char in mapping.values():
                stack.append(char)
            elif char in mapping.keys():
                if not stack or mapping[char] != stack.pop():
                    counter -= 1

    return counter

isValid(["((()))","(()())","(())()","()(())","()()()"])
```
