```python


def extract_images_pypdf(pdf_object,page_number_start,page_number_end):
    reader = pdf_object
    image_count = 0
    image_list = []
    if page_number_start == page_number_end:
        for page_num, page in enumerate(reader.pages[page_number_start]):
            print(page_num)
            image_temp = []
            try:
                for img_file_ref in page.images:
                    # Decode the image data
                    image_data = img_file_ref.data
                    image_ext = img_file_ref.ext

                    # Create a PIL Image object from the bytes
                    image = Image.open(io.BytesIO(image_data))
                    
                    # Save the image
                    #image_filename = f"{output_folder}/page{page_num+1}_image{image_count+1}.{image_ext}"
                    #image.save(image_filename)
                    #print(f"Saved: {image_filename}")
                    image_count += 1
                    image_temp.append(image)
                image_list.append(image_temp)
            except Exception as e:
                print(f"Could not extract image on page {page_num+1}: {e}")
                continue
    else:
        for page_num, page in enumerate(reader.pages[page_number_start:page_number_end]):
            image_temp = []
            try:
                for img_file_ref in page.images:
                    # Decode the image data
                    image_data = img_file_ref.data
                    image_ext = img_file_ref.ext

                    # Create a PIL Image object from the bytes
                    image = Image.open(io.BytesIO(image_data))
                    
                    # Save the image
                    #image_filename = f"{output_folder}/page{page_num+1}_image{image_count+1}.{image_ext}"
                    #image.save(image_filename)
                    #print(f"Saved: {image_filename}")
                    image_count += 1
                    image_temp.append(image)
                image_list.append(image_temp)
            except Exception as e:
                print(f"Could not extract image on page {page_num+1}: {e}")
                continue
                
    print(f"\nExtraction complete. Total images saved: {image_count}")
    return image_list


```

```python
# Usage: Replace 'your_document.pdf' with your PDF file name
image_list = extract_images_pypdf(fearon_pdf_object, 4,4)
```
